### Separate Numerical Vs Categorical Features

```python
num_feature = [feature for feature in df.columns if df[feature].dtype == np.float64]
cat_feature = [feature for feature in df.columns if df[feature].dtype not in [np.int64, np.float64]]
```

### Remove Outliers

```python
plt.figure(figsize=(6, 4))
sns.boxplot(data=df)
plt.xticks(rotation=60)
plt.suptitle("Outlier Checking", fontsize = 14, fontweight='bold')
plt.xlabel("Feature Name")
plt.ylabel("Counts")
plt.show()
```

```python
def return_bounds(df, feature):
    q1 = df[feature].quantile(0.25)
    q3 = df[feature].quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.25 * iqr
    uppder_bound = q3 + 1.25 * iqr
    return lower_bound, uppder_bound
```

```python
from feature_engine.outliers import ArbitraryOutlierCapper
for feature in num_feature:
    lower_bound, upper_bound = return_bounds(df, feature)
    arb = ArbitraryOutlierCapper(min_capping_dict={feature: lower_bound}, max_capping_dict={feature: upper_bound})
    df = arb.fit_transform(df)
```

### Train Test Split

```python
x = df.drop('Class', axis=1)
y = df['Class']
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42, stratify=y)
```

```python
# print(df['Class'].unique())
print(df['Class'].value_counts())
```

### Preprocessing

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

## Machie Learning Performance

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
import xgboost as xg
from xgboost import XGBClassifier
```

```python
lr = LogisticRegression()
rf = RandomForestClassifier(random_state=23, n_estimators=100)
dt = DecisionTreeClassifier(max_depth=10)
svm = SVC(probability=True)
xg = XGBClassifier(n_estimators=100)
```

```python
lr.fit(x_train_scaled, y_train)
svm.fit(x_train_scaled, y_train)
xg.fit(x_train_scaled, y_train)
dt.fit(x_train_scaled, y_train)
rf.fit(x_train_scaled, y_train)
```

```python
y_pred1 = lr.predict(x_test_scaled)
y_pred2 = svm.predict(x_test_scaled)
y_pred3 = xg.predict(x_test_scaled)
y_pred4 = dt.predict(x_test_scaled)
y_pred5 = rf.predict(x_test_scaled)
```

### Model Performance

```python
from sklearn.metrics import accuracy_score
print("Logistic Regression : ", accuracy_score(y_test, y_pred1))
print("\nSVM : ", accuracy_score(y_test, y_pred2))
print("\nXGBoost : ", accuracy_score(y_test, y_pred3))
print("\nDecision Tree : ", accuracy_score(y_test, y_pred4))
print("\nRandom Forest : ", accuracy_score(y_test, y_pred5))
```

## Deep Learning

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Dropout, BatchNormalization
from tensorflow.keras.callbacks import EarlyStopping
```

```python
model = Sequential()

model.add(Dense(128, activation='relu', input_dim = 30))
model.add(Dropout(0.1))
model.add(BatchNormalization())
model.add(Dense(64, activation='relu'))
model.add(Dropout(0.1))
model.add(BatchNormalization())
model.add(Dense(32, activation='relu'))
model.add(Dropout(0.1))
model.add(BatchNormalization())

# Output Layer
model.add(Dense(1, activation='sigmoid'))
```

### Model Compilation

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

## EarlyStoping

```python
callback = EarlyStopping(
    monitor = 'val_loss',
    min_delta = 0.0001,
    patience = 20,
    verbose = 1,
    mode = 'auto',
    baseline = None,
    restore_best_weights = False
)
```

```python
history = model.fit(x_train_scaled, y_train, epochs = 100, validation_data=(x_test_scaled, y_test), callbacks=[callback])
```

### Accuracy

```python
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.legend()
plt.show()
```

### Loss

```python
plt.plot(history.history['loss'], label='Traning Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()
plt.show()
```
